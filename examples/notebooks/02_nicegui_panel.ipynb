# NiceGUI dentro do notebook

Use o kernel **Neuronal — Local** desta branch. NiceGUI executa no mesmo
processo do kernel e o painel aparece em um iframe na saída da célula.
JupyterLab e navegador devem estar na mesma máquina; Jupyter remoto/HTTPS
precisa de proxy e ainda não é suportado por este protótipo.

O painel pertence ao pacote; este notebook é uma das formas de apresentação.
O mesmo painel abre com `python examples/panel.py`.


```python
from neuronal.panel import TimeAxisPanel, shutdown_panels

# Executar novamente fecha o painel anterior e cria outro experimento.
if "panel" in globals():
    panel.close()
panel = TimeAxisPanel(samples=1000, dt_ms=0.1)
await panel.show()
```

Ajuste os controles e clique em **Calculate**. Execute a célula abaixo para ler o estado atualizado; outputs de células anteriores não se atualizam automaticamente.

```python
panel.parameters, panel.result[:10], len(panel.result)
```

Também podemos alterar os parâmetros em Python e reapresentar o painel.

```python
panel.parameters.update(samples=200, dt_ms=0.5)
_ = panel.compute()
panel.close()
panel = TimeAxisPanel(**panel.parameters)
await panel.show()
```

Feche o painel quando terminar. O servidor permanece disponível para outros painéis até desligar o kernel.

```python
panel.close()
```

Para encerrar o servidor: `await shutdown_panels()`. Depois, reinicie o kernel antes de criar novos painéis.
